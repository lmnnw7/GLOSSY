# Digital-core FID using a pretrained 3D ResNet

Set the paths to your real cores, generated cores and pretrained 3D ResNet checkpoint, then run the cells in order.

Use the GLOSSY environment created from `GLOSSY.yaml` in the project root.
Activate it before launching Jupyter; no separate dependency installation is required.

```python
import sys
from pathlib import Path

PROJECT_DIR = Path.cwd() if (Path.cwd() / "fid_3d.py").exists() else Path.cwd() / "evaluation" / "digital_core"
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

from fid_3d import compute_fid_3d

REAL_DIR = r"../../data/real_binary"        # Real digital-core directory
# FAKE_DIR = r"../../data/generated_binary"          # Generated digital-core directory
FAKE_DIR = r"../../data/ablations/wo-dit-b"          # Generated digital-core directory
CHECKPOINT = r"../../checkpoints/resnet_50_23dataset.pth"  # Set the path to your pretrained 3D ResNet checkpoint

# RAW/DAT/BIN/IMG inputs require shape and dtype; set shape=None for NPY/NPZ inputs.
SHAPE = (128, 128, 128)
DTYPE = "int8"
```

```python
result = compute_fid_3d(
    real=REAL_DIR,
    fake=FAKE_DIR,
    checkpoint=CHECKPOINT,
    depth=50,
    shape=SHAPE,
    dtype=DTYPE,
    resize=(128, 128, 128),
    batch_size=2,
    device="auto",
    output="fid_dcb.json",
)

result
```

```python
# Use this loop to compare outputs from multiple generative models.
fake_groups = {
    "dcb": r"../../data/baselines/dcgan",
    "style": r"../../data/baselines/stylegan",
    "st": r"../../data/baselines/stgan",
    "geb": r"../../data/generated_binary",
}

all_results = {}
for name, folder in fake_groups.items():
    all_results[name] = compute_fid_3d(
        real=REAL_DIR,
        fake=folder,
        checkpoint=CHECKPOINT,
        depth=50,
        shape=SHAPE,
        dtype=DTYPE,
        resize=(128, 128, 128),
        batch_size=2,
        device="auto",
        output=f"fid_{name}.json",
    )

all_results
```
